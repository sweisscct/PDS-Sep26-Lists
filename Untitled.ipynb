```python
lecurers = ["Sam", "David", "Taufique", "Muhammad"]
new_element_position = 3
first_half = lecurers[:new_element_position]
new_element = "Martha"
second_half = lecurers[new_element_position:]

lecurers = first_half
lecurers.append(new_element)
lecurers.extend(second_half)
print(lecurers)
```

## lists as iterables
- we can use a for loop directly on the list elements

```python
for lecturer in lecurers:
    if lecturer == "Sam":
        print(f"I love {lecturer}'s class")
    else:
        print(f"I like {lecturer}'s class")
print("And that's all my classes!")
```

## list comprehension
we can make a list directly from a for loop

```python
# [element_value for element in iterable]
square_numbers = [num**2 for num in range(10)]
```

```python
square_numbers
```

```python
len(square_numbers)
```

## `enumerate` 
This lets us iterate through a list, while also keeping track of each index

```python
for lecturer in lecurers:
    print(lecturer)
```

```python
for index, lecturer in enumerate(lecurers):
    print(f"{index+1}: {lecturer}")
```

## Dictionaries
- Useful for assosiating varaible with a (unique) name
- Work with key=value pairs
    - Access data via the key, not the index
    - Order is not gaurenteed
- (like maps or JSON in other languages)

```python
modules_map = {"Programming": ["Sam", "Thursday"], "Data Vis": ["David", "Tuesday"]}
```
